```python
#Trapezoidal rule
#S=h((f(a)+f(b))/2 + f(x1)+ f(x2)+..........+f(xn-1)
def f(x):
    return (1/(1+x))
n=int(input("Enter no. of intervals: "))
a=float(input("Enter lower: "))
b=float(input("Enter upper: "))
h=(b-a)/n
s=(0.5*(f(a)+f(b)))
for i in range (1,n):
    s+=f(a+i*h)
s*=h
print(s)
```

```python
#Simpsons 1/3
def f(x):
    return (1/(1+x))
n=int(input("Enter no. of intervals: "))
a=float(input("Enter lower: "))
b=float(input("Enter upper: "))
h=(b-a)/n
s=f(a)+f(b)
for i in range(1,n):
    if(i%2==0):
        s+=2*f(a+i*h)
    else:
        s+=4*f(a+i*h)
s=(s*h)/3
print(s)
```

```python
#bisection method
def f(x):
    return (x**3-x-11)
x0=float(input("Enter starting point: "))
x1=float(input("Enter ending point: "))
e=float(input("Enter tolerance: "))
if (f(x0)*f(x1)>0):
    print("Wrong interval given")
else:    
    x=(x0+x1)/2
    err=x1-x0
    i=0
    while err>=e:
        if(f(x)==0):
            print(f"root = {x},f{x}={f(x)}")
        if (f(x)>0):
            x1=x
        else:
            x0=x
        err=x1-x0
        x=(x0+x1)/2
        i+=1
    print(f'Approximate root = {x}, f({x})={f(x)})')
```

```python
#langrange interpolation
fi=[]
xi=[]
s=0
n=int(input("Enter n: "))
x=float(input("Enter x: "))
for i in range(0,n):
    valx=float(input(f'Enter x[{i+1}]: '))
    xi.append(valx)
    valff=float(input(f'Enter corresponding values of function value f[{i+1}]: '))
    fi.append(valff)
s=0
for i in range(0,n):
    a=1
    for j in range(0,n):
        if (j!=i):
            a*=((x-xi[j])/(xi[i]-xi[j]))
    s+=fi[i]*a
print(f'for {x} value of function is {s}')
```
